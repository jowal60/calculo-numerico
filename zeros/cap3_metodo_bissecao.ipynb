# Método da Bisseção

Este notebook apresenta o **método da bisseção**, um dos algoritmos mais simples e robustos para encontrar raízes de equações não lineares.

O método exige:
- uma função contínua \( f(x) \)
- um intervalo inicial \([a,b]\) tal que \( f(a)f(b) < 0 \)

A cada iteração, o intervalo é reduzido pela metade até que a raiz seja aproximada com a tolerância desejada.

## Gráfico da Função

Antes de aplicar o método, é útil visualizar a função para confirmar a existência de uma raiz no intervalo.

```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: x**3 - x - 2

x = np.linspace(-2, 3, 400)
y = f(x)

plt.figure(figsize=(8,4))
plt.axhline(0, color='black', linewidth=1)
plt.plot(x, y, label='f(x) = x³ - x - 2')
plt.grid(True)
plt.legend()
plt.title('Gráfico da Função')
plt.show()
```

## Implementação do Método da Bisseção

A função abaixo implementa o método e imprime a tabela de iterações:

```python
def bissecao(f, a, b, tol=1e-6, max_iter=100):
    print("| k | a | b | pm | b - a | f(pm) |")
    print("|---|---|---|----|--------|--------|")

    if f(a) * f(b) >= 0:
        raise ValueError("O método da bisseção requer f(a)*f(b) < 0.")

    for k in range(1, max_iter + 1):
        pm = (a + b) / 2
        intervalo = b - a
        fpm = f(pm)

        print(f"| {k} | {a:.6f} | {b:.6f} | {pm:.6f} | {intervalo:.6f} | {fpm:.6e} |")

        if abs(fpm) < tol or intervalo/2 < tol:
            return pm, k

        if f(a) * fpm < 0:
            b = pm
        else:
            a = pm

    return pm, max_iter
```

## Exemplo de Uso

```python
raiz, iteracoes = bissecao(f, 1, 2)
print("\nRaiz aproximada:", raiz)
print("Iterações:", iteracoes)
```

## Conclusão

O método da bisseção é simples, robusto e garante convergência sempre que a função muda de sinal no intervalo inicial.
